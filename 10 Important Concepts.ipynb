```python
class Car:
    def __init__(self, brand: str, horsepower: int) -> None:
        self.brand = brand
        self.horsepower = horsepower

    def drive(self) -> None:
        print(f"{self.brand} is driving")

    def get_information(self) -> None:
        print(f"{self.brand} is with {self.horsepower} horsepower")

Volvo = Car("Volvo", 2000)
Volvo.drive()
Volvo.get_information()

bmw = Car("BMW", 25000)
bmw.drive()
bmw.get_information()
```

```python
class Car:
    def __init__(self, brand: str, horsepower: int) -> None:
        self.brand = brand
        self.horsepower = horsepower

    def drive(self) -> None:
        print(f"{self.brand} is driving")

    def get_information(self, var: int) -> None:
        print(var)
        print(f"{self.brand} is with {self.horsepower} horsepower")

Volvo = Car("Volvo", 2000)
Volvo.drive()
Volvo.get_information(150)

bmw = Car("BMW", 25000)
bmw.drive()
bmw.get_information(250)
```

```python
class Car:
    def __init__(self, brand: str, horsepower: int) -> None:
        self.brand = brand
        self.horsepower = horsepower

    def __str__(self) -> str:
        return f"{self.brand}, {self.horsepower} hp"

    def __add__(self, other: Self) -> str:
        return f'{self.brand} & {other.brand}'

volvo = Car("Volvo", 20500)
bmw = Car("BMW", 3000)
print(volvo)
print(volvo + bmw)
```
